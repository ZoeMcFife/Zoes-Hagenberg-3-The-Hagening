## Structure of ipynb
Notebooks consist of cells. These could be either *markdown*, *code* or *result* cells.<br>

**markdown** Text-Cells; used for structuring, documentation and notes. Uses markdown syntax [https://www.markdownguide.org/basic-syntax].<br>
⚠️ <font color="gray">Not all editors/viewers support everything or render certain things different.</font><br>
⚠️ <font color="gray">Be careful with colors! Some use</font> <font color="white">dark</font> <font color="gray">mode, others prefer</font> <font color="black">light</font> <font color="gray">backgrounds. The effect could be very different.</font><br>
⚠️ <font color="gray">When the cell keeps showing 'source-text', hit [crtrl]+[enter] to render it (VSCode - might work differently in your editor). To edit, simply double-click it.</font> <br>

**code** Code-Cells; used for programming, executing code, generating output.<br>
⚠️ <font color="gray">Notebook code execution is not neccessary linear! Code cells can be executed in any order. Depending on the code, this can have significant impact on the result.</font><br>
⚠️ <font color="gray">The notebook is an _interactive_ means of interfacing a python kernel. Code will read current values of data or variables in the kernel, manipulate them and store new values. These changes will be present as long as the kernel is running.</font><br>
⚠️ <font color="gray">Make sure your notebooks 'works' from top to bottom in a single run from a fresh kernel! Use '*restart*' and '*run all*' (might be labeled differently in your editor) frequently - at least _before_ you upload your work.</font><br>

**result** Output-Cells; used for displaying results from executing code-cells.<br>
⚠️ <font color="gray">Not all editors/viewers support everything or render certain things differently! You may have to install additional plugins/extensions to make it work on your own.</font><br>
