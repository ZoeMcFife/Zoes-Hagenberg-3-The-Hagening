## Data types

```python
#Columns are in German and names are inconsistent
df = df.rename( columns={"name": "Name", "PLZ": "ZIP", "land": "Country", "bundesland":"State",
                    "bezirk":"District", "stadt": "City", "strasse":"Address", "datum":"Date"} )
```

```python
#ZIP-Code is read as a number, but we don't do calculations on it, it's better threated as an string so no statistics is calculated on it:
df.ZIP = df.ZIP.astype(str)
```

```python
pd.DataFrame({"is:":df.dtypes,"should be:":["str", "float64", "float64", "str", "str", "str", "str", "str", "str","datetime65[us]", "float64", "float64"]})
```

```python
# Fuel prices are still strings, let's have a look at the data
df.loc[:10,["Super95","Diesel"]]
```

```python
# remove 'EUR' -> replace ' EUR' with ''
# This is not working -> 'replace' works only for whole string, not parts
df[['Super95', 'Diesel']] = df[['Super95','Diesel']].replace(' EUR', '')

# Need to add 'regex=True' to replace substrings
df[['Super95', 'Diesel']] = df[['Super95', 'Diesel']].replace(' EUR', '', regex=True)
```

```python
#what about those 1.7990000000000002? -> float precision errors, save to cut after 3 decimal digits (5 characters with 1.123)
df['Super95'] = df['Super95'].str[:5]
df['Diesel'] = df['Diesel'].str[:5]
```

```python
# Fuel prices are still strings, let's have a look at the last 3 rows of fuel data
print(df.iloc[-3:][["Super95","Diesel"]], "\n\n", df[["Super95","Diesel"]].dtypes)
```

```python
#after import, no automatic conversion, so need to explicitly be converted:
df[['Super95','Diesel']] = df[['Super95','Diesel']].astype(float)
df[['Super95','Diesel']].dtypes
```

```python
# Convert dates
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)
pd.DataFrame({"is:":df.dtypes,"should be:":["str", "float64", "float64", "str", "str", "str", "str", "str", "str","datetime64[us]", "float64", "float64"]})
```

## Data cleaning

### Data capitalization

```python
# Different capitalization in e.g. Cities / Addresses:
df.loc[[4, 5, 44, 106, 107, 300]]
```

### Cleanup of Names

```python
df["Name"] = df["Name"].str.title()
df["Country"] = df["Country"].str.title()
df["State"] = df["State"].str.title()
df["District"] = df["District"].str.title()
df["City"] = df["City"].str.title()
df["Address"] = df["Address"].str.title()
```

```python
df.loc[[4, 5, 44, 106, 107, 300]]
```

```python
for col in df.columns:
    if df[col].dtype == "str":
        df[col] = df[col].replace(" An Der ", " a. d. ", regex=True)
        df[col] = df[col].replace(" A. D. ", " a. d. ", regex=True)
        df[col] = df[col].replace(" Im ", " im ", regex=True)
        df[col] = df[col].replace(" In ", " in ", regex=True)
        df[col] = df[col].replace(" Bei ", " bei ", regex=True)
        df[col] = df[col].replace(" B. ", " bei ", regex=True)
        df[col] = df[col].replace("Sankt ", "St.", regex=True)
        df[col] = df[col].replace(r"St\. ", "St.", regex=True)
        df[col] = df[col].replace("Wr. Neu", "Wiener Neu", regex=True)
        df[col] = df[col].replace("Puch/Hallein", "Puch bei Hallein")
        df[col] = df[col].replace("Puch-Hallein", "Puch bei Hallein")
        df[col] = df[col].replace("Waidhofen/Thaya", "Waidhofen a. d. Thaya")
        df[col] = df[col].replace("Bruck / Leitha", "Bruck a. d. Leitha")
        df[col] = df[col].replace("Bruck/Leitha", "Bruck a. d. Leitha")
        df[col] = df[col].replace("Brueckl", "Brückl")
        df[col] = df[col].replace("D-Ehrensdorf", "Deutsch Ehrensdorf")
        df[col] = df[col].replace("Deutsch-Griffen", "Deutsch Griffen")
        df[col] = df[col].replace("Deutsch-Wagram" , "Deutsch Wagram")
        
        
```

```python
df["Name"] = df["Name"].replace("Bp", "BP")
df["Name"] = df["Name"].replace("Diskont-Tankstelle", "Diskont Tankstelle", regex=True)
```

```python
df["Address"] = df["Address"].replace("Am Hofer Parkplatz", "Hofer Pp.", regex=True)
df["Address"] = df["Address"].replace("Hofer Parkplatz", "Hofer Pp.", regex=True)
df["Address"] = df["Address"].replace("Hofer-Parkplatz", "Hofer Pp.", regex=True)
```

```python
df["Address"] = df["Address"].replace(r"str\.", "straße", regex=True)
df["Address"] = df["Address"].replace(r"Str\.", "Straße", regex=True)
df["Address"] = df["Address"].replace("trasse", "traße", regex=True)
df["Address"] = df["Address"].replace("staße", "straße", regex=True)
df["Address"] = df["Address"].replace("traße;", "traße", regex=True)
df["Address"] = df["Address"].replace("Vis A Vis", "vav.", regex=True)
```
