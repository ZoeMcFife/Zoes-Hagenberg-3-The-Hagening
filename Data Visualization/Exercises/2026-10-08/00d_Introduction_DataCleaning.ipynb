### Extracting more Infos - derived data

```python
# Find all full-self-service gas stations
df['Automat'] = df['Name'].str.contains('utomat')

#create new column with on-premise gas stations (Hofer, Lagerhaus)
# 1. Define the conditions
conditions = [
    df['Address'].str.contains('Hofer Pp.', na=False),
    df['Name'].str.contains('agerhaus', na=False),
    df['Address'].str.contains('Lagerhaus ', na=False),
    df['Name'].str.contains('Genol', na=False)
]

# 2. Define the corresponding values
choices = ['Hofer', 'Lagerhaus', 'Lagerhaus', 'Lagerhaus']

# 3. Create the new column (assign 'Unknown' or None if no match)
df['OnPrem'] = np.select(conditions, choices, default='Unknown')
```
